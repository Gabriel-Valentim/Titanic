## 3. Análise de Valores Faltantes

```python
# Função para analisar valores faltantes
def missing_values_analysis(df, dataset_name):
    """
    Analisa e visualiza valores faltantes no dataset
    """
    print(f"\n🔍 Análise de Valores Faltantes - {dataset_name}")
    print("="*60)
    
    # Calcular valores faltantes
    missing = df.isnull().sum()
    missing_pct = (missing / len(df)) * 100
    
    # Criar DataFrame com resultados
    missing_df = pd.DataFrame({
        'Coluna': missing.index,
        'Valores Faltantes': missing.values,
        'Percentual (%)': missing_pct.values
    })
    
    # Filtrar apenas colunas com valores faltantes
    missing_df = missing_df[missing_df['Valores Faltantes'] > 0].sort_values(
        'Valores Faltantes', ascending=False
    )
    
    if len(missing_df) > 0:
        print(missing_df.to_string(index=False))
        return missing_df
    else:
        print("✅ Nenhum valor faltante encontrado!")
        return None

# Analisar train e test
missing_train = missing_values_analysis(train, "TREINO")
missing_test = missing_values_analysis(test, "TESTE")
```

```python
# Visualizar valores faltantes no dataset de treino
if missing_train is not None:
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))
    
    # Gráfico de barras - quantidade
    ax[0].bar(missing_train['Coluna'], missing_train['Valores Faltantes'], color='salmon')
    ax[0].set_title('Valores Faltantes por Coluna (Quantidade)', fontsize=14, fontweight='bold')
    ax[0].set_ylabel('Quantidade')
    ax[0].tick_params(axis='x', rotation=45)
    
    # Gráfico de barras - percentual
    ax[1].bar(missing_train['Coluna'], missing_train['Percentual (%)'], color='lightcoral')
    ax[1].set_title('Valores Faltantes por Coluna (Percentual)', fontsize=14, fontweight='bold')
    ax[1].set_ylabel('Percentual (%)')
    ax[1].tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()
```

## 4. Análise da Variável Target (Survived)

```python
# Distribuição da variável target
print("🎯 Distribuição da Variável Target (Survived):")
print("="*60)
survived_counts = train['Survived'].value_counts()
survived_pct = train['Survived'].value_counts(normalize=True) * 100

result_df = pd.DataFrame({
    'Survived': ['Não (0)', 'Sim (1)'],
    'Quantidade': survived_counts.values,
    'Percentual (%)': survived_pct.values.round(2)
})

print(result_df.to_string(index=False))
print(f"\n📊 Taxa de Sobrevivência: {survived_pct[1]:.2f}%")
```

```python
# Visualizar distribuição do target
fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico de barras
colors = ['#FF6B6B', '#4ECDC4']
ax[0].bar(['Não Sobreviveu', 'Sobreviveu'], survived_counts.values, color=colors)
ax[0].set_title('Distribuição de Sobrevivência (Quantidade)', fontsize=14, fontweight='bold')
ax[0].set_ylabel('Quantidade de Passageiros')
for i, v in enumerate(survived_counts.values):
    ax[0].text(i, v + 10, str(v), ha='center', fontweight='bold')

# Gráfico de pizza
ax[1].pie(survived_counts.values, labels=['Não Sobreviveu', 'Sobreviveu'], 
          autopct='%1.1f%%', colors=colors, startangle=90)
ax[1].set_title('Proporção de Sobrevivência', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
